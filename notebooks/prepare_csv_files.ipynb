# Load libraries

```python
import pandas as pd
```

# Read cleaned files

```python
data = pd.read_csv('../data/clean/stores_cleaned.csv')
data.head()
```

## Tables without foreign keys

### Countries

```python
countries = pd.DataFrame({"country_id": [i+1 for i in range(data['country'].nunique())], "country": [value if value != '中国' else 'China' for value in data['country'].unique()]})
countries
```

```python
countries.to_csv("../data/clean/countries.csv", index=False, sep=";", encoding="utf-8")
```

### Payment methods

```python
data = pd.read_csv('../data/clean/transactions_cleaned.csv')
data.head()
```

```python
#create dictionary for id key and name
# with list comprehension check nunique()

payment_methods = pd.DataFrame({"payment_method_id": [i+1 for i in range(data['payment_method'].nunique())], "payment_method": [value for value in data['payment_method'].unique()]})
payment_methods
```

```python
payment_methods.to_csv("../data/clean/payment_methods.csv", index=False, sep=";", encoding="utf-8")
```
