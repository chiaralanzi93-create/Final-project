# Data cleaning

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df_customers = pd.read_csv('../data/raw/customers.csv', low_memory=False)
df_discounts = pd.read_csv('../data/raw/discounts.csv', low_memory=False)
df_employees = pd.read_csv('../data/raw/employees.csv', low_memory=False)
df_products = pd.read_csv('../data/raw/products.csv', low_memory=False)
df_stores = pd.read_csv('../data/raw/stores.csv', low_memory=False)
df_transactions = pd.read_csv('../data/raw/transactions.csv', low_memory=False)
```

**Customer table**

```python
df_customers.head()
```

```python
df_customers.info()
```

```python
df_customers.isnull().sum()
```

```python
df_customers['Job Title'] = df_customers['Job Title'].fillna(' ')
```

```python
df_customers['Gender'].unique()
```

```python
df_customers.columns = df_customers.columns.str.lower().str.replace(' ', '_')
```

**Discount table**

```python
df_discounts.head()
```

```python
df_discounts.info()
```

```python
df_discounts.columns = df_discounts.columns.str.lower().str.replace(' ', '_')
```

```python
df_discounts['category'].unique()
```

```python
df_discounts['sub_category'].unique()
```

```python
df_discounts['start'] = pd.to_datetime(df_discounts['start']) # convert from object to datetime
df_discounts['end'] = pd.to_datetime(df_discounts['end'])
```

```python
df_discounts['discont']=df_discounts['discont']/ 100
```

**Product table**

```python
df_products.head()
```

```python
df_products = df_products.drop(columns=['Description ES','Description FR','Description DE','Description PT','Description ZH'])
```

```python
df_products.columns = df_products.columns.str.lower().str.replace(' ', '_')
```

```python
df_products['category'].unique()
```

```python
df_products['sub_category'].unique()
```

```python
df_products
```

**Store table**

```python
df_stores.info()
```

```python
df_stores.columns = df_stores.columns.str.lower().str.replace(' ', '_')
```

**Transaction table**

```python
df_transactions.head()
```

```python
df_transactions.columns = df_transactions.columns.str.lower().str.replace(' ', '_')
```

```python
df_transactions.info()
```

```python
df_transactions['date'] = pd.to_datetime(df_transactions['date']) # convert from object to datetime
```

```python
df_transactions['Month'] = df_transactions['date'].dt.month #new column
df_transactions['Year'] = df_transactions['date'].dt.year
df_transactions['Day'] = df_transactions['date'].dt.dayofweek
```

```python
df_transactions.isna().sum() # not relevant for business focus 
```

## Data wrangling - unique fact table

```python
df= df_transactions.merge(df_products, on = 'product_id', how = 'left')
df
```

```python
df = df.merge(df_stores, on = 'store_id', how = 'left')
df
```

```python
df = df.drop(['size','color_x','color_y','sizes'],axis=1)
```

```python

df.to_csv('../data/clean/Data_cleaning_wrangling.csv',index=False)
```
